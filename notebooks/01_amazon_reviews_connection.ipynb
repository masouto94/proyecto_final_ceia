# Conexión a Amazon Reviews'23 (HuggingFace)

Kickstart de HU1: carga del dataset `McAuley-Lab/Amazon-Reviews-2023` desde HuggingFace Hub, inspirado en https://amazon-reviews-2023.github.io/data_loading/huggingface.html.

**Nota:** la guía oficial usa `load_dataset("McAuley-Lab/Amazon-Reviews-2023", "raw_review_All_Beauty", trust_remote_code=True)`, pero ese repo depende de un *loading script* (`Amazon-Reviews-2023.py`) que las versiones actuales de `datasets` (>=3.0) ya no soportan (`Dataset scripts are no longer supported`). Como alternativa equivalente, se cargan directamente los archivos `.jsonl` crudos del repo vía el esquema `hf://datasets/...`, que es la forma que hoy recomienda HuggingFace para este tipo de datasets.

Se usa una sola categoría (`All_Beauty`) a modo de prueba de conexión, ya que el dataset completo pesa cientos de GB.

```python
from datasets import load_dataset
import pandas as pd
```

```python
# Categoría chica para validar la conexión; el listado completo de categorías está en all_categories.txt del repo
CATEGORY = "All_Beauty"

reviews = load_dataset(
    "json",
    data_files=f"hf://datasets/McAuley-Lab/Amazon-Reviews-2023/raw/review_categories/{CATEGORY}.jsonl",
    split="train",
)
reviews
```

```python
df = reviews.to_pandas()
df.head()
```
